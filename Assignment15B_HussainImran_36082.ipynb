Q2: What is the average number of cups per gender?

```python
st.groupby("gender")["cups_per_day"].mean()
```

Q3: Make a bar chart of snack choices.

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(data=st, x='snack_choice')

plt.xlabel('Snack Choice')
plt.ylabel('Number of People')
plt.title('Distribution of Snack Choices')
plt.show()
```
